# Классификация ориентации текстового кропа (0° / 180°)
Решение: обучение на синтетических данных, классификатор MobileNetV3-Small,
калибровка вероятностей, формирование файла предсказаний.
Конвейер воспроизводится с нуля; ручная разметка не применяется.

```python
import os, io, glob, json, time
import numpy as np
import pandas as pd
from PIL import Image, ImageDraw, ImageFont, ImageFilter, ImageEnhance
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

# Фиксация случайных зерен для воспроизводимости результатов
torch.manual_seed(42)
np.random.seed(42)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('device =', device)
```

## Формирование набора шрифтов
- открытые кириллические шрифты Google Fonts и системные шрифты. При отсутствии сетевого доступа используются только системные.

```python
os.makedirs('fonts', exist_ok=True)
urls = [
 'https://github.com/google/fonts/raw/main/ofl/ptsans/PT_Sans-Web-Regular.ttf',
 'https://github.com/google/fonts/raw/main/ofl/ptsans/PT_Sans-Web-Bold.ttf',
 'https://github.com/google/fonts/raw/main/ofl/neucha/Neucha-Regular.ttf',
 'https://github.com/google/fonts/raw/main/ofl/badscript/BadScript-Regular.ttf',
]

import urllib.request


for u in urls:
    fn = 'fonts/' + u.split('/')[-1]
    if not os.path.exists(fn):
        try:
            urllib.request.urlretrieve(u, fn)
        except Exception as e:
            print('не удалось загрузить', fn, e)

def ok_cyr(path):
    # Проверка наличия кириллических глифов: рендеринг символов кириллицы
    # сравнивается с рендерингом неназначенных кодов Unicode (символ подстановки)
    try:
        f = ImageFont.truetype(path, 24)
    except Exception:
        return False
    def px(s):
        im = Image.new('L', (220, 44), 0)
        ImageDraw.Draw(im).text((2, 2), s, font=f, fill=255)
        return np.asarray(im).sum()
    a = px('ЫыЪъЭэБбДЛФ')
    b = px('\u0378\u0379\u037a')
    return a > 0 and a != b

cand = []
cand += glob.glob('fonts/*.ttf')
for d in ['/usr/share/fonts', '/usr/local/share/fonts', '/System/Library/Fonts',
          os.path.expanduser('~/.fonts'), os.path.expanduser('~/.local/share/fonts'),
          'C:/Windows/Fonts']:
    for e in ('ttf', 'otf', 'ttc'):
        cand += glob.glob(d + '/**/*.' + e, recursive=True)
try:
    import matplotlib
    cand += glob.glob(os.path.join(matplotlib.get_data_path(), 'fonts', 'ttf', 'DejaVu*.ttf'))
except Exception:
    pass

fonts = []
for p in cand:
    if os.path.isfile(p) and ok_cyr(p) and p not in fonts:
        fonts.append(p)
print('найдено шрифтов с кириллицей:', len(fonts))
assert len(fonts) > 0
for f in fonts[:6]:
    print('  ', f)
```

- #### Словарь соответствует целевому домену (тексты объявлений).
- #### Генератор строк воспроизводит типичные формы: верхний регистр, цены, телефонные номера, строки длиной от одного до шести слов.

```python
WORDS = ('товар цена продам куплю квартира комната кухня диван стол стул шкаф кровать '
'телефон айфон самсунг ноутбук компьютер планшет телевизор холодильник машина автомобиль '
'колеса диски куртка ботинки кроссовки платье рубашка джинсы шапка сумка часы кольцо серьги '
'браслет игрушка коляска велосипед самокат лыжи палатка гараж дача участок дом баня недорого '
'срочно новый гарантия доставка обмен торг звоните пишите самовывоз этаж район улица метро '
'год пробег комплект набор коробка документы чек ремонт установка монтаж оплата карта '
'наличные скидка акция распродажа подарок оригинал размер цвет белый черный красный синий '
'зеленый серый коричневый услуга сервис мастер замена').split()


def gen_text(r):
    n = r.integers(1, 7)
    w = [str(r.choice(WORDS)) for _ in range(n)]
    x = r.random()
    if x < 0.15:
        s = ' '.join(w).upper()
    elif x < 0.25:
        s = w[0].capitalize()
        if n > 1:
            s += ' ' + ' '.join(w[1:])
    elif x < 0.35:
        s = '%d %s' % (r.integers(100, 99999), r.choice(WORDS))
    elif x < 0.42:
        s = '%d-%d-%d' % (r.integers(1000, 9999), r.integers(10, 99), r.integers(10, 99))
    else:
        s = ' '.join(w)
    return s
```

## Синтез изображений

```python
def make_bg(r, w, h):
    k = r.integers(0, 4)
    if k == 0:    # Сплошная заливка
        c = r.integers(0, 256, 3).astype(np.uint8)
        return np.tile(c, (h, w, 1))
    if k == 1:    # Линейный градиент
        t = np.linspace(0, 1, w)[None, :, None]
        a = r.integers(0,256,3)[None,None,:]*(1-t) + r.integers(0,256,3)[None,None,:]*t
        return np.broadcast_to(a, (h, w, 3)).astype(np.uint8)
    if k == 2:    # Гладкие цветовые пятна (имитация расфокусированного фона)
        sm = r.integers(0, 256, (4, 4, 3)).astype(np.uint8)
        return np.array(Image.fromarray(sm).resize((w, h), Image.BICUBIC))
    base = r.integers(0, 256, 3).astype(np.float32)    # Аддитивный шум.
    return np.clip(base + r.normal(0, 25, (h, w, 3)), 0, 255).astype(np.uint8)

_fcache = {}
def get_font(path, size):
    # Кэш объектов шрифта для снижения накладных расходов рендеринга
    if (path, size) not in _fcache:
        _fcache[(path, size)] = ImageFont.truetype(path, size)
    return _fcache[(path, size)]

def make_crop(r):
    # Рендеринг текстовой строки на сформированном фоне
    text = gen_text(r)
    fp = fonts[int(r.integers(0, len(fonts)))]
    fnt = get_font(fp, int(r.integers(26, 60)))
    pad = int(r.integers(4, 24))

    tmp = ImageDraw.Draw(Image.new('RGB', (8, 8)))
    l, t, rr, b = tmp.textbbox((0, 0), text, font=fnt)
    tw, th = max(4, rr - l), max(4, b - t)
    W, H = tw + 2*pad, th + 2*pad

    bg = make_bg(r, W, H)
    im = Image.fromarray(bg)
    dr = ImageDraw.Draw(im)
    if bg.mean() > 127:
        fg = (0, 0, 0) if r.random() < 0.2 else tuple(int(v) for v in r.integers(0, 60, 3))
    else:
        fg = (255, 255, 255) if r.random() < 0.2 else tuple(int(v) for v in r.integers(200, 256, 3))
    dr.text((pad - l, pad - t), text, font=fnt, fill=fg)

    # Имитация искажений, характерных для боксов детектора: наклон, перспектива, снижение разрешения, размытие, изменение яркости и контраста, сжатие JPEG
    if r.random() < 0.5:
        im = im.rotate(float(r.uniform(-10, 10)), resample=Image.BICUBIC,
                       fillcolor=tuple(int(v) for v in bg[0, 0]))
    if r.random() < 0.4:
        w, h = im.size
        q = []
        for (qx, qy) in [(0, 0), (0, h), (w, h), (w, 0)]:
            q += [qx + r.uniform(-0.06, 0.06)*w, qy + r.uniform(-0.06, 0.06)*h]
        im = im.transform((w, h), Image.QUAD, tuple(q), Image.BICUBIC)
    if r.random() < 0.5:
        w, h = im.size
        s = float(r.uniform(0.35, 0.7))
        im = im.resize((max(8, int(w*s)), max(8, int(h*s))), Image.BILINEAR).resize((w, h), Image.BILINEAR)
    if r.random() < 0.6:
        im = im.filter(ImageFilter.GaussianBlur(float(r.uniform(0.3, 1.4))))
    im = ImageEnhance.Brightness(im).enhance(float(r.uniform(0.7, 1.3)))
    im = ImageEnhance.Contrast(im).enhance(float(r.uniform(0.7, 1.4)))
    if r.random() < 0.7:
        buf = io.BytesIO()
        im.save(buf, 'JPEG', quality=int(r.integers(45, 96)))
        buf.seek(0)
        im = Image.open(buf).convert('RGB')

    # Поворот на 180 градусов выполняется последним и определяет метку класса
    y = 0
    if r.random() < 0.5:
        im = im.rotate(180)
        y = 1
    return im, y
```

## Предобработка и датасеты

```python
IH, IW = 64, 256
tot = transforms.ToTensor()
norm = transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

def prep(im):
    # Масштабирование с сохранением пропорций; недостающая ширина заполняется дублированием краевого столбца. Нормализация соответствует статистике ImageNet
    w, h = im.size
    nw = int(round(w * IH / h))
    nw = max(8, min(IW, nw))
    im = im.resize((nw, IH), Image.BILINEAR)
    a = np.array(im, np.uint8)
    if nw < IW:
        a = np.pad(a, ((0, 0), (0, IW - nw), (0, 0)), mode='edge')
    return norm(tot(Image.fromarray(a)))

class SynthDS(Dataset):
    # Датасет генерируется на обращении: элемент с индексом i детерминирован парой (seed, i), что обеспечивает воспроизводимость без хранения файлов
    def __init__(self, n, seed):
        self.n = n
        self.seed = seed
    def __len__(self):
        return self.n
    def __getitem__(self, i):
        r = np.random.default_rng([self.seed, i])
        im, y = make_crop(r)
        return prep(im), np.float32(y)

class FilesDS(Dataset):
    def __init__(self, paths):
        self.paths = paths
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, i):
        return prep(Image.open(self.paths[i]).convert('RGB'))
```

## Модель
- Базовая сеть — MobileNetV3-Small (преобучение на ImageNet): минимальное число параметров при достаточной точности
-  Альтернатива — облегченная CNN, обучаемая с нуля

```python
def make_model(kind='mv3'):
    if kind == 'mv3':
        m = models.mobilenet_v3_small(weights=models.MobileNet_V3_Small_Weights.DEFAULT)
        inf = m.classifier[0].in_features
        m.classifier = nn.Sequential(
            nn.Linear(inf, 512), nn.Hardswish(), nn.Dropout(0.2), nn.Linear(512, 1))
    else:
        m = nn.Sequential(
            nn.Conv2d(3, 16, 3, 2, 1), nn.BatchNorm2d(16), nn.ReLU(),
            nn.Conv2d(16, 32, 3, 2, 1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 64, 3, 2, 1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 128, 3, 2, 1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 8)), nn.Flatten(),
            nn.Linear(1024, 64), nn.ReLU(), nn.Linear(64, 1))
    return m.to(device)
```

## Обучение
- Бинарная кросс-энтропия на логитах, оптимизатор AdamW, расписание OneCycleLR
- Качество на валидации контролируется метрикой Brier и долей верных ответов

```python
BS, EPOCHS = 128, 10
tr = DataLoader(SynthDS(30000, 1234), batch_size=BS, shuffle=True,
                num_workers=2, pin_memory=True, drop_last=True)
va = DataLoader(SynthDS(4000, 777), batch_size=256, num_workers=2)
# При проблемах с мультипроцессингом в Windows следует задать num_workers=0

net = make_model('mv3')
opt = torch.optim.AdamW(net.parameters(), lr=3e-4, weight_decay=1e-4)
sch = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=1e-3, epochs=EPOCHS, steps_per_epoch=len(tr))
lossf = nn.BCEWithLogitsLoss()

def eval_brier(net, dl):
    # Оценка качества на валидации: метрика Brier и доля верных ответов
    net.eval()
    P, Y = [], []
    with torch.no_grad():
        for x, y in dl:
            P.append(torch.sigmoid(net(x.to(device))).squeeze(-1).cpu())
            Y.append(y)
    p = torch.cat(P).numpy()
    y = torch.cat(Y).numpy()
    return float(((p - y) ** 2).mean()), float(((p > .5) == (y > .5)).mean())

for ep in range(EPOCHS):
    net.train()
    s = 0
    c = 0
    for x, y in tqdm(tr, desc='ep %d' % ep):
        x = x.to(device)
        y = y.to(device)
        opt.zero_grad(set_to_none=True)
        loss = lossf(net(x).squeeze(-1), y)
        loss.backward()
        opt.step()
        sch.step()
        s += loss.item() * len(x)
        c += len(x)
    br, ac = eval_brier(net, va)
    print('ep', ep, 'loss %.4f' % (s / c), 'val brier %.4f' % br, 'acc %.4f' % ac)
```

## Калибровка вероятностей
- Температурное масштабирование: параметр T подбирается перебором сетки на валидации
по минимуму отрицательного логарифма правдоподобия, что снижает избыточную уверенность предсказаний
- Калибровка необходима, поскольку метрика Brier штрафует избыточную уверенность

```python
net.eval()
LG, YY = [], []
with torch.no_grad():
    for x, y in va:
        LG.append(net(x.to(device)).squeeze(-1).cpu())
        YY.append(y)
lg = torch.cat(LG)
yy = torch.cat(YY)

bestT, bestNll = 1.0, 1e9
for T in np.linspace(0.3, 3.0, 55):
    nll = F.binary_cross_entropy_with_logits(lg / T, yy).item()
    if nll < bestNll:
        bestNll = nll
        bestT = float(T)
print('T =', bestT)
```

## Формирование файла предсказаний
При отсутствии тестового каталога генерируется демонстрационный набор с известными метками.
Вероятности вычисляются с калибровочной температурой, ограничиваются диапазоном [0.01, 0.99]
и записываются в submission.csv.

```python
TEST_DIR = r'../../test/images'    # Каталог с тестовыми кропами
EXT = ('.jpg', '.jpeg', '.png', '.webp', '.bmp')

if not os.path.isdir(TEST_DIR) or len(glob.glob(TEST_DIR + '/*')) == 0:
    print('тестовый каталог отсутствует, формируется демонстрационный набор')
    os.makedirs(TEST_DIR, exist_ok=True)
    r = np.random.default_rng(2024)
    demo = {}
    for i in range(1000):
        im, y = make_crop(r)
        nm = 'box_%06d.jpg' % i
        im.save(TEST_DIR + '/' + nm)
        demo[nm] = y
    json.dump(demo, open('demo_labels.json', 'w'))

files = [f for f in sorted(glob.glob(TEST_DIR + '/*')) if f.lower().endswith(EXT)]
print('число тестовых кропов:', len(files))

net.eval()
pr = []
with torch.no_grad():
    for x in DataLoader(FilesDS(files), batch_size=256, num_workers=2):
        pr.append(torch.sigmoid(net(x.to(device)) / bestT).squeeze(-1).cpu())
p = torch.cat(pr).numpy()

# Ограничение диапазона [0.01, 0.99] исключает катастрофический вклад ошибочных предсказаний с предельной уверенностью в метрику Brier.
sub = pd.DataFrame({'image_id': [os.path.basename(f) for f in files],
                    'p_180': np.clip(p, 0.01, 0.99)})
sub.to_csv(r'../../sample_submission.csv', index=False)
sub.head()

if os.path.exists('demo_labels.json'):
    dm = json.load(open('demo_labels.json'))
    if set(sub.image_id) <= set(dm):
        y = np.array([dm[n] for n in sub.image_id], np.float32)
        print('demo 1-brier = %.4f' % (1 - ((sub.p_180.values - y) ** 2).mean()))
```
